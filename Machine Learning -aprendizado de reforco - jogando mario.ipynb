```python
import time
import gym
import gym_super_mario_bros
from nes_py.wrappers import JoypadSpace
from gym_super_mario_bros.actions import RIGHT_ONLY

env = gym_super_mario_bros.make('SuperMarioBros-v1')
env = JoypadSpace(env, RIGHT_ONLY)
obs = env.reset()

for i in range(1000):
    env.render()
    time.sleep(0.02)
    acao = 1
    step_result = env.step(acao)

    if len(step_result) == 5:
        obs, recompensa, terminou, truncated, info = step_result
        terminou = terminou or truncated
    elif len(step_result) == 4:
        obs, recompensa, terminou, info = step_result
    else:
        raise ValueError("Número inesperado de valores retornados pela função step")

    print(recompensa)
    if terminou:
        print(f"Episódio finalizado depois de {i+1} steps")
        break

env.close()
```

```python
import gym_super_mario_bros
from nes_py.wrappers import JoypadSpace
from gym_super_mario_bros.actions import RIGHT_ONLY
import time
from OpenAIWrappers import nosso_wrapper
from stable_baselines3.common.vec_env import SubprocVecEnv
from stable_baselines3 import PPO
from stable_baselines3.common.callbacks import CheckpointCallback
from gym.wrappers.monitoring import video_recorder


def cria_env(nome_env):
    def _init():
        env = gym_super_mario_bros.make(nome_env)
        env = JoypadSpace(env, RIGHT_ONLY)
        env = nosso_wrapper(env)
        return env
    return _init


threads_cpu = 2
env = SubprocVecEnv([cria_env(nome_env='SuperMarioBros-v1') for i in range(threads_cpu)])

checkpoint_callback = CheckpointCallback(save_freq=6250, save_path='C:/Users/linco/Downloads/modelos/', name_prefix='n1024b64l3')

modelo = PPO(policy='CnnPolicy', env=env, gamma=0.099, n_steps=512, batch_size=32, learning_rate=0.0003, vf_coef=0.5, ent_coef=0.01, verbose=1)

inicio = time.time()
modelo.learn(total_timesteps=10000, log_interval=10, callback=checkpoint_callback)

fim = time.time()
print('Tempo total do treinamento em horas:', (fim-inicio)/3600)
```
